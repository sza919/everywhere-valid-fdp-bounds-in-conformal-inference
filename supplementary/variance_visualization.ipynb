```python
import numpy as np
import matplotlib.pyplot as plt
fs = 14
plt.rcParams.update({
    "font.size": fs,
    "axes.titlesize": fs,
    "axes.labelsize": fs,
    "xtick.labelsize": fs,
    "ytick.labelsize": fs,
    "legend.fontsize": fs
})
def rho_n(t, n):
    """
    rho_n(t) in Var(H_t) = rho_n(t) * t * (1 - t).

    t can be a scalar or numpy array in (0, 1).
    """
    t = np.asarray(t, dtype=float)
    N = n + 1
    a = N * t
    k = np.floor(a).astype(int)
    gamma = a - k

    numerator = N * (k + gamma**2) - a**2
    denominator = (n + 2) * a * (N - a)

    return numerator / denominator

# Avoid endpoints because rho_n(t) is written as a ratio with t(1-t) in denominator.
t = np.linspace(1e-4, 1 - 1e-4, 4000)

plt.figure(figsize=(8, 5.5))

for n in [2, 5, 10, 30, 100]:
    plt.plot(t, rho_n(t, n), label=f"n = {n}")

plt.xlabel("t")
plt.ylabel(r"$\rho_n(t)$")
plt.title(r"Plot of $\rho_n(t)$ for several values of $n$")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()

# Save figure
plt.savefig("rho_n(t).pdf", bbox_inches="tight")

plt.show()
```
